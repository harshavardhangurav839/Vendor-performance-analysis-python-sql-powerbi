Now we are going to make Vendor_sales_summary table which as data of freight table which we previously made along with table as purchase summary and sales summary so after that we joined the tables by it's uniqueness vendor number and brand.

```python

Vendor_sales_summary = pd.read_sql_query("""
    WITH freight_data AS (
        SELECT VendorNumber, SUM(Freight) AS freight_cost
        FROM vendor_invoice
        GROUP BY VendorNumber
    ),
    PurchaseSummary AS (
        SELECT 
            p.VendorNumber, p.VendorName, p.Brand,
            p.PurchasePrice, pp.Volume,
            pp.Price AS actual_price,
            SUM(p.Quantity) AS TotalPurchaseQuantity,
            SUM(p.Dollars)  AS TotalPurchaseDollars
        FROM purchases p
        JOIN purchase_prices pp ON p.Brand = pp.Brand
        WHERE p.PurchasePrice > 0
        GROUP BY p.VendorNumber, p.VendorName, p.Brand
    ),
    SalesSummary AS (
        SELECT 
            VendorNo, Brand,
            Description,
            SUM(SalesDollars)  AS TotalSalesDollars,
            SUM(SalesPrice)    AS TotalSalesPrice,
            SUM(SalesQuantity) AS TotalSalesQuantity,
            SUM(ExciseTax)     AS TotalExciseTax
        FROM sales
        GROUP BY VendorNo, Brand, Description
    )
    SELECT
        ps.VendorNumber, ps.VendorName, ps.Brand,
        ps.PurchasePrice, ps.actual_price AS ActualPrice,
        ps.Volume, ps.TotalPurchaseQuantity, ps.TotalPurchaseDollars,
        ss.TotalSalesQuantity, ss.TotalSalesDollars,
        ss.TotalSalesPrice, ss.TotalExciseTax,
        ss.Description,
        fd.freight_cost
    FROM PurchaseSummary ps
    LEFT JOIN SalesSummary ss 
           ON ps.VendorNumber = ss.VendorNo AND ps.Brand = ss.Brand
    LEFT JOIN freight_data fd 
           ON ps.VendorNumber = fd.VendorNumber
    ORDER BY ps.TotalPurchaseDollars DESC
""", conn)

print("DataFrame created:", Vendor_sales_summary.shape)
print(Vendor_sales_summary.columns.tolist())                                                                                                                                                                                                                                      


                                 
```

Now we got the final table name as vendor_sales_summary on which we are going to be make analysis so before moving towards it, we need to make ensure that data inside  is in clean form for that we use fuctions like unique(),fillna(),str.strip() etc . In data cleaning process we erase null and duplicate values, convert into columns into appropriate datatypes, replace the particular data , remove spaces from string datatypes.

We also need to add columns inside the table which will become useful for further analysis such gross profit,profit margin,stockturnover and sales to purchase ratio.

```python
Vendor_sales_summary['GrossProfit']=Vendor_sales_summary['TotalSalesDollars']-Vendor_sales_summary['TotalPurchaseDollars']
```

```python
Vendor_sales_summary['ProfitMargin']=Vendor_sales_summary['GrossProfit']/Vendor_sales_summary['TotalSalesDollars']*100
```

```python
Vendor_sales_summary['StockTurnover']=Vendor_sales_summary['TotalSalesQuantity']/Vendor_sales_summary['TotalPurchaseQuantity']
```

```python
Vendor_sales_summary['SalesToPurchaseRatio']=Vendor_sales_summary['TotalSalesDollars']/Vendor_sales_summary['TotalPurchaseDollars']
```
